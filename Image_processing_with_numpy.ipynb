```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# load image and conert to rgb
img = Image.open("sample.png").convert("RGB")
print(img)

# convert img into numpy array 
image = np.array(img)

# image anlysis
print("Image Shape:", image.shape)
print("Image Dimention:", image.ndim)
print("Data Type:", image.dtype)
print("Minimium Pixel:", image.min())
print("Maximium Pixel:", image.max())
print("Average Pixel:", image.mean())

plt.figure(figsize=(10,6)) #plt.figure image ka canvas batata ha figsize = canvas ki width 10 insches and height 6 insches rakhana 

plt.imshow(image)# image ka display show krta ha 
plt.axis("off")# axis =  side ky no and axis ko hide krta ha 
plt.show() # image ko visulize krta ha
```
